```python
import torch
from models.hr_mamba import HRMambaRegressor

import warnings 
import pandas as pd
warnings.filterwarnings("ignore")
ckpt_path = "HR_reg_MAMBA_v1.mdl"
device = "cuda:0" if torch.cuda.is_available() else "cpu" 

downstream_model  = HRMambaRegressor(
    input_channels=3,
    seq_len=300,
    patch_size=15,
    stride=15,
    d_model=128,
    depth=4,
    d_state=16,
).float().to(device)

pretrained = torch.load(ckpt_path, map_location=device)

# remove a head antiga do pretraining

model_dict = downstream_model.state_dict()
model_dict.update({k: v for k, v in pretrained.items() if k in model_dict})
downstream_model.load_state_dict(model_dict)

print("Backbone carregado; nova head downstream mantida.")
```
